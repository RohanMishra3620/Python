```python
df["tokens"]=df["Review"].apply(word_tokenize)
```

```python

df[["Review", "tokens"]].head()
df["tokens"] = df["tokens"].apply(
    lambda tokens: [word for word in tokens if word.isalnum()]
)

df[["Review", "tokens"]].head()
```

```python
stop=set(stopwords.words("English"))
stop = set(stopwords.words("english"))

stop.remove("not")
stop.remove("no")
stop.remove("nor")
```

```python
df["tokens"]=df["tokens"].apply(lambda token : [word for word in token if word not in stop])
```
