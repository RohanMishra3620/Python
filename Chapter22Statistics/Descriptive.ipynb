```python
import numpy as np
data=np.random.randn(1000)
mean=np.mean(data)
n=len(data)
num=np.sum((data-mean)**3)/n
dem=(np.sum((data-mean)**2)/n)**(3/2)
skewness=num/dem
print("Mean",mean)
print("Skeness",skewness)
```

```python
import numpy as np
from scipy.stats import skew
import matplotlib.pyplot as plt 
print(skew(np.random.randn(1000)))
plt.hist((np.random.randn(1000)),bins=30)
plt.show()
```
