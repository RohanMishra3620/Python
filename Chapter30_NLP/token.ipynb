```python
import nltk
# nltk.download('punkt')
from nltk.tokenize import word_tokenize ,sent_tokenize
```

```python
sent=sent_tokenize("Natural Language Processing (NLP) is a branch of artificial intelligence that enables computers to understand, interpret, and generate human language. It is used in applications such as chatbots, language translation, sentiment analysis, speech recognition, and text summarization. By combining linguistics and machine learning, NLP helps machines communicate with people in a more natural and meaningful way.")
for i, sentence in enumerate(sent, 1):
    print("Sentence", i, ":", sentence)
```

```python
tokens=word_tokenize("Every day is a new opportunity to learn something, grow, and become a better version of yourself.")
tokens
```

```python
from nltk.corpus import stopwords
nltk.download("stopwords")
```

```python
stopwords=stopwords.words("English")
print(len(stopwords))
print(stopwords)
```

```python
filtered=[word for word in tokens if word.lower() not in stopwords]
print(filtered)
```

```python

from nltk.stem import PorterStemmer
steammer=PorterStemmer()
print(steammer.stem("running"))
print(steammer.stem("flying"))
print(steammer.stem("better"))
print(steammer.stem("careful"))
print(steammer.stem("careing"))
print(steammer.stem("universities"))
```

```python
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
leammatizer=WordNetLemmatizer()
print(leammatizer.lemmatize("running" ,pos="v"))
print(leammatizer.lemmatize("flying",pos="v"))
print(leammatizer.lemmatize("better",pos="a"))
print(leammatizer.lemmatize("careful",pos="a"))
print(leammatizer.lemmatize("universities",pos="n"))
```
