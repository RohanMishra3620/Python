```python
import scipy.stats as stats
n=80
sample_mean=130
sample_std=40
mu_0=120 
```

step 1: Set up the hypotheses

```python
t_statistic=(sample_mean-mu_0)/(sample_std/(n**0.5))
t_statistic
```

```python
p_value=1-stats.t.cdf(t_statistic,df=n-1)
p_value
```

```python
if p_value< 0.05:
    print("reject")
else :
    print("accept")
```
