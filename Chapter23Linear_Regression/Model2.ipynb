Linear Regression Overview

Linear Regression is a supervised learning algorithm used to model the relationship between one or more independent variables (features) and a continuous dependent variable (target).

Types of Linear Regression:
1. Simple (or Single) Linear Regression:

Uses one independent variable to predict a continuous target variable by fitting a straight line.

2. Multiple Linear Regression:

Uses two or more independent variables to predict the target by fitting a multi-dimensional hyperplane.




Predicting Sales Based on Store Footfalls using Simple Linear Regression

Neeche explanation likha hai:

We want to predict store sales based on footfall using a simple linear regression model.

```python
import pandas as pd 
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
data=pd.read_csv("General_Stores.csv")
data
```

```python
data.corr()
```

we use x=data[["Footfalls"]] because in sllearn independent as Dataframe

```python
x=data[["Footfalls"]]
y=data["DailySales"]
```

```python
plt.xlabel("FootFalls")
plt.ylabel("Daliysales")
plt.scatter(x,y,color="red",marker="o")
```

```python
model=LinearRegression()
model.fit(x,y)
```

```python
print(f"Cofficient (slope): {model.coef_}")
print(f"Intercept: {model.intercept_}")
```

calulate what will be prediction value for footfall of 20

```python
print(model.intercept_ +model.coef_*20)
print(model.predict([[20]]))
#we caheck with both method and it is same with maths 
# b1+b0*c and also with predict both are same
```

calulate what will be prediction value for footfall of 35

```python
model.predict([[35]])
```

```python
# predict the value
new = pd.DataFrame({"Footfalls": [45, 50, 100]})

predicted_sales = model.predict(new)

print(pd.DataFrame({
    "Footfalls": new["Footfalls"],
    "Predicted_sales": predicted_sales
}))
```
